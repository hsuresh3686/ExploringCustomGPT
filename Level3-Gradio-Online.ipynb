```python
from openai import OpenAI
client = OpenAI()
import gradio
```

```python
messages = [{"role": "system", "content" : "You are a Tomcat Manager Assistant specializes in all type of request related to Tomcat"}]

def CustomGPT(user_input):
    messages.append({"role":"user", "content": user_input})sages=messages)
    response = client.responses.create(
        model="gpt-5-mini",
        input=messages)
    reply = response.output_text
    messages.append({"role":"assistant", "context": reply})
    return reply

demo = gradio.Interface(fn=CustomGPT, inputs="text", outputs = "text", title="Tomcat Assistant")
demo.launch(share=True)
```

```python
demo.close()
```
